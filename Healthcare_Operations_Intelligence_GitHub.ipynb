# 11. Synthetic Data-Generation Audit: Symptom to Department

The most important data-quality insight is not a missing value; it is the structure of the synthetic data itself. The symptom feature is extremely associated with department. Cramer's V is 0.969, and a simple symptom-to-most-common-department lookup reaches 95.63% accuracy.

This means a model that uses symptoms could obtain very high accuracy by learning a relationship that is already almost hard-coded in the synthetic data. The enhanced project therefore treats symptoms as an audit target and excludes it from the final benchmark. This is a methodological decision to keep the ML evaluation educationally meaningful.

*Figure 10. Strong synthetic symptom-to-department association.*

## Dataset check

````python
# 2) Locate the source CSV

expected_files = [
    DATA_DIR / "healthcare_data.csv",
    DATA_DIR / "healthcare_raw.csv",
    DATA_DIR / "healthcare.csv",
]

DATA_PATH = next((p for p in expected_files if p.exists()), None)

# Also allow a single CSV inside data/ without forcing a file rename.
if DATA_PATH is None:
    csv_candidates = sorted(DATA_DIR.glob("*.csv"))
    if len(csv_candidates) == 1:
        DATA_PATH = csv_candidates[0]

DATA_AVAILABLE = DATA_PATH is not None

if DATA_AVAILABLE:
    print("Using:", DATA_PATH)
else:
    print(
        "No healthcare CSV was found in data/. "
        "The report itself did not include the raw dataset, so data-dependent cells are skipped until it is added."
    )
````

````python
# 3) Expected schema and small helpers

EXPECTED_COLUMNS = [
    "patient_id",
    "age",
    "gender",
    "symptoms",
    "temperature",
    "duration_days",
    "bp_systolic",
    "bp_diastolic",
    "heart_rate",
    "oxygen_saturation",
    "pain_level",
    "symptom_severity",
    "chronic_condition",
    "department",
]

def require_columns(frame, columns):
    missing = [c for c in columns if c not in frame.columns]
    if missing:
        raise ValueError(
            "The CSV is missing required columns: " + ", ".join(missing)
        )

def cramers_v(x, y):
    table = pd.crosstab(x, y)
    if table.empty:
        return np.nan
    chi2 = chi2_contingency(table)[0]
    n = table.to_numpy().sum()
    r, k = table.shape
    if n <= 1:
        return np.nan

    phi2 = chi2 / n
    phi2corr = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1))
    rcorr = r - ((r - 1) ** 2) / (n - 1)
    kcorr = k - ((k - 1) ** 2) / (n - 1)
    denominator = min(kcorr - 1, rcorr - 1)

    return np.sqrt(phi2corr / denominator) if denominator > 0 else np.nan

def minmax(series):
    lo = series.min()
    hi = series.max()
    if hi == lo:
        return pd.Series(np.zeros(len(series)), index=series.index)
    return (series - lo) / (hi - lo)
````

````python
# 4) Load and audit the raw data

if DATA_AVAILABLE:
    df_raw = pd.read_csv(DATA_PATH)
    require_columns(df_raw, EXPECTED_COLUMNS)

    print("Raw shape:", df_raw.shape)
    print("Duplicate rows:", int(df_raw.duplicated().sum()))
    print("Missing cells:", int(df_raw.isna().sum().sum()))

    audit = pd.DataFrame({
        "dtype": df_raw.dtypes.astype(str),
        "missing_count": df_raw.isna().sum(),
        "missing_pct": (df_raw.isna().mean() * 100).round(3),
        "unique_count": df_raw.nunique(dropna=False),
    })
    display(audit)
else:
    df_raw = None
````

````python
# 5) Cleaning pipeline from the report

if DATA_AVAILABLE:
    df = df_raw.copy()

    # Preserve the raw dataframe; cleaning happens on a copy.
    df["_had_missing"] = df.isna().any(axis=1)

    duplicate_mask = df.duplicated(keep="first")
    duplicate_count = int(duplicate_mask.sum())
    df = df.loc[~duplicate_mask].copy()

    categorical_cols = [
        "gender",
        "symptoms",
        "symptom_severity",
        "chronic_condition",
        "department",
    ]

    for col in categorical_cols:
        df[col] = df[col].astype("string").str.strip()

    # Keep ENT as an acronym.
    dept_normalize = {
        "ent": "ENT",
        "E.N.T.": "ENT",
        "e.n.t.": "ENT",
    }
    df["department"] = df["department"].replace(dept_normalize)

    # Cap oxygen saturation values above 100.
    oxygen_over_100 = int((df["oxygen_saturation"] > 100).sum())
    df.loc[df["oxygen_saturation"] > 100, "oxygen_saturation"] = 100.0

    # Treat extreme systolic BP values as missing before imputation.
    extreme_bp_mask = (df["bp_systolic"] < 60) | (df["bp_systolic"] > 200)
    extreme_bp_count = int(extreme_bp_mask.sum())
    df.loc[extreme_bp_mask, "bp_systolic"] = np.nan

    numeric_cols = [
        "age",
        "temperature",
        "duration_days",
        "bp_systolic",
        "bp_diastolic",
        "heart_rate",
        "oxygen_saturation",
        "pain_level",
    ]

    for col in numeric_cols:
        df[col] = df[col].fillna(df[col].median())

    for col in categorical_cols:
        mode = df[col].mode(dropna=True)
        if not mode.empty:
            df[col] = df[col].fillna(mode.iloc[0])

    df["quality_status"] = np.where(
        df["_had_missing"],
        "required_correction",
        "clean",
    )

    df = df.drop(columns=["_had_missing"])

    print("Rows after duplicate removal:", len(df))
    print("Final missing cells:", int(df.isna().sum().sum()))
    print("Unique patient IDs:", df["patient_id"].nunique())

    df.to_csv(DATA_DIR / "healthcare_cleaned_150000.csv", index=False)
    df.to_csv(DATA_DIR / "healthcare_enriched_150000.csv", index=False)
else:
    df = None
````

## Department Operations Intelligence — executable calculation

````python
# 8) Operations Load Index
# Report weights:
# volume share 40%
# severe-case share 25%
# chronic-condition share 15%
# average symptom duration 20%

if DATA_AVAILABLE:
    dept = df.groupby("department").agg(
        patients=("department", "size"),
        severe_pct=(
            "symptom_severity",
            lambda s: (s.astype(str).str.lower() == "severe").mean() * 100,
        ),
        chronic_pct=(
            "chronic_condition",
            lambda s: (s.astype(str).str.lower() == "yes").mean() * 100,
        ),
        avg_duration=("duration_days", "mean"),
    ).reset_index()

    dept["volume_pct"] = dept["patients"] / len(df) * 100

    dept["volume_norm"] = minmax(dept["volume_pct"])
    dept["severe_norm"] = minmax(dept["severe_pct"])
    dept["chronic_norm"] = minmax(dept["chronic_pct"])
    dept["duration_norm"] = minmax(dept["avg_duration"])

    dept["ops_load"] = (
        0.40 * dept["volume_norm"]
        + 0.25 * dept["severe_norm"]
        + 0.15 * dept["chronic_norm"]
        + 0.20 * dept["duration_norm"]
    ) * 100

    dept = dept.sort_values("ops_load", ascending=False)

    display(
        dept[
            [
                "department",
                "patients",
                "volume_pct",
                "severe_pct",
                "chronic_pct",
                "avg_duration",
                "ops_load",
            ]
        ]
    )

    dept.to_csv(ARTIFACT_DIR / "department_operations_load.csv", index=False)
else:
    dept = None
````

````python
# 9) Operations Load Index chart
if DATA_AVAILABLE:
    plt.figure(figsize=(10, 5))
    plt.bar(dept["department"], dept["ops_load"])
    plt.title("Operations Load Index by Department")
    plt.xlabel("Department")
    plt.ylabel("Index (0-100)")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.savefig(CHART_DIR / "operations_load_index.png", dpi=150)
    plt.show()
````

## Synthetic data-generation audit — symptom to department

````python
# 10) Quantify the symptom -> department relationship
if DATA_AVAILABLE:
    cv = cramers_v(df["symptoms"], df["department"])

    symptom_lookup = (
        df.groupby("symptoms")["department"]
        .agg(lambda s: s.mode().iloc[0])
    )
    lookup_pred = df["symptoms"].map(symptom_lookup)
    lookup_accuracy = (lookup_pred == df["department"]).mean()

    print("Cramer's V:", round(cv, 3))
    print("Symptom lookup accuracy:", round(lookup_accuracy * 100, 2), "%")

    leakage = {
        "cramers_v": float(cv),
        "symptom_lookup_accuracy": float(lookup_accuracy),
    }
    (ARTIFACT_DIR / "synthetic_leakage_audit.json").write_text(
        json.dumps(leakage, indent=2),
        encoding="utf-8",
    )
````

## Exploratory Machine Learning

````python
# 11) Department classification benchmark

if DATA_AVAILABLE:
    target = "department"
    features = [
        "age",
        "temperature",
        "duration_days",
        "bp_systolic",
        "bp_diastolic",
        "heart_rate",
        "oxygen_saturation",
        "pain_level",
        "gender",
        "chronic_condition",
        "symptom_severity",
    ]

    X = df[features].copy()
    y = df[target].copy()

    numeric_features = [
        "age",
        "temperature",
        "duration_days",
        "bp_systolic",
        "bp_diastolic",
        "heart_rate",
        "oxygen_saturation",
        "pain_level",
    ]
    categorical_features = [
        "gender",
        "chronic_condition",
        "symptom_severity",
    ]

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "num",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="median")),
                    ("scaler", StandardScaler()),
                ]),
                numeric_features,
            ),
            (
                "cat",
                Pipeline([
                    ("imputer", SimpleImputer(strategy="most_frequent")),
                    ("onehot", OneHotEncoder(handle_unknown="ignore")),
                ]),
                categorical_features,
            ),
        ]
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        stratify=y,
        random_state=RANDOM_STATE,
    )

    models = {
        "Majority Baseline": DummyClassifier(strategy="most_frequent"),
        "Logistic Regression": LogisticRegression(max_iter=2000),
        "Decision Tree": DecisionTreeClassifier(random_state=RANDOM_STATE),
        "Random Forest": RandomForestClassifier(
            n_estimators=200,
            random_state=RANDOM_STATE,
            n_jobs=-1,
        ),
    }

    results = []
    trained = {}

    for name, model in models.items():
        pipe = Pipeline([
            ("preprocessor", preprocessor),
            ("model", model),
        ])

        pipe.fit(X_train, y_train)
        pred = pipe.predict(X_test)

        results.append({
            "Model": name,
            "Accuracy": accuracy_score(y_test, pred),
            "Balanced accuracy": balanced_accuracy_score(y_test, pred),
            "Macro F1": f1_score(y_test, pred, average="macro"),
        })
        trained[name] = pipe

    benchmark = pd.DataFrame(results).sort_values("Accuracy", ascending=False)
    display(benchmark)
    benchmark.to_csv(ARTIFACT_DIR / "model_benchmark.csv", index=False)
else:
    benchmark = None
````

````python
# 12) ML benchmark chart
if DATA_AVAILABLE:
    chart_data = benchmark.set_index("Model")[[
        "Accuracy",
        "Balanced accuracy",
        "Macro F1",
    ]]

    ax = chart_data.plot(kind="bar", figsize=(10, 5))
    ax.set_title("Four-Model Exploratory Benchmark")
    ax.set_ylabel("Score")
    ax.set_xlabel("Model")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.savefig(CHART_DIR / "four_model_benchmark.png", dpi=150)
    plt.show()
````
