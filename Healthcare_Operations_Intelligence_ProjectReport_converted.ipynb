# 3. Problem Statement

Healthcare facilities can generate large amounts of structured patient information. Without a systematic analysis workflow, patterns in demographics, symptoms, department workloads, vital-sign measurements, severity labels and chronic-condition fields can be difficult to inspect consistently. The project therefore asks: how can a synthetic patient dataset be converted into a transparent, reusable analytical product that surfaces operational patterns, validates data quality and evaluates whether non-symptom features contain useful predictive signal for department assignment?

The project does not attempt to diagnose patients or recommend treatment.

# 10. Department Operations Intelligence

The enhanced project introduces an Operations Load Index. It is a custom analytical score designed to summarize operational dimensions in one transparent view. It combines four components: patient volume share (40%), severe-case share (25%), chronic-condition share (15%) and average symptom duration (20%). Each component is min-max normalized across departments before the weighted sum. The score is not a medical severity or risk score.

# 11. Synthetic Data-Generation Audit: Symptom to Department

The most important data-quality insight is not a missing value; it is the structure of the synthetic data itself. The symptom feature is extremely associated with department. Cramer's V is 0.969, and a simple symptom-to-most-common-department lookup reaches 95.63% accuracy.

This means a model that uses symptoms could obtain very high accuracy by learning a relationship that is already almost hard-coded in the synthetic data. The enhanced project therefore treats symptoms as an audit target and excludes it from the final benchmark. This is a methodological decision to keep the ML evaluation educationally meaningful.

Figure 10. Strong synthetic symptom-to-department association.

## 12.1 Baseline and enhanced model benchmark

The enhanced benchmark's strongest model by accuracy is Random Forest at 22.49% accuracy and 21.35% balanced accuracy. Although this improves on the majority baseline and the single Decision Tree, the absolute performance remains modest. The result is therefore reported as an exploratory ML finding rather than a clinical prediction capability.

Figure 11. Four-model exploratory benchmark.

The original baseline report's Decision Tree value (16.83%) and the enhanced reproducible benchmark's Decision Tree value (16.79%) are very close. Small differences can arise from implementation details such as preprocessing and encoding, so the enhanced project emphasizes relative comparison, baselines and transparent methodology rather than claiming a large model-performance leap.

The Streamlit application turns the analysis into an interactive product. Users can navigate between executive KPIs, data quality, department intelligence, patient/symptom exploration, ML audit, evidence-based Q&A and downloads.

Evidence Q&A - fixed evidence-based answers calculated directly from the cleaned dataset.

Downloads - cleaned/enriched CSVs and analytical artifacts.

Because Streamlit is not required for the notebook itself, the project remains usable in a notebook-only environment while the dashboard provides the presentation layer for a final submission or demonstration.

# 16. Limitations

The dataset is synthetic and cannot be generalized to real patient populations.

There is no time dimension, so longitudinal trends and per-day workload cannot be inferred.

The symptom-to-department relationship is unusually strong because of synthetic data generation.

The cleaned dataset relies on median/mode imputation for missing values.

The final ML task is department classification, not a clinical outcome prediction.

The Operations Load Index is a custom descriptive score, not a clinical risk or severity score.

The project does not contain diagnoses, lab results, treatment outcomes, hospital stay duration, medication history or imaging data.

No external clinical validation has been performed.

# 18. Conclusion

The enhanced Healthcare Operations Intelligence project demonstrates a complete, reproducible analytics workflow on a synthetic healthcare dataset. It preserves the baseline project's important findings - including 150,000 cleaned records, the dominance of Dermatology in volume, the prominence of Chest Pain, the 37.28% chronic-condition prevalence and the age-group patterns - while adding stronger engineering and analytical layers.

The most important methodological improvement is the explicit audit of synthetic feature structure. With Cramer's V = 0.969 and symptom lookup accuracy of 95.63%, the project avoids presenting an artificially high symptom-based model as genuine predictive performance. The enhanced model benchmark then provides a more honest comparison among multiple models using features independent of the strongest synthetic shortcut.

Overall, the project is positioned as a data analytics and operational intelligence demonstration - not as a medical decision system.
